```python
sns.lineplot(x=df['Intervention_Response'] , y=df['Reward_Score'])
plt.show()
```

### The line plot clearly defines the relation between the Intervention Response and Reward Score.

```python
plt.figure(figsize=(10,6))
corr_cols = ['Study_Hours','Exam_Pressure','Stress_Score','Anxiety_Score']
corr_matrix = df[corr_cols].corr()
sns.heatmap(corr_matrix , annot=True)
plt.show()
```

##

### The above HeatMap shows the Relation Between features 'Study_Hours','Exam_Pressure','Stress_Score','Anxiety_Score'.
